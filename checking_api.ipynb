```python
import requests 
import json
```

```python
url = "http://213.173.99.48:29046/generate"
```

```python
question = """
Qönçə Nəzərli tərəfindən sistemdəki qeydiyyatı yaradılmış, lakin 
qeydiyyat ünvanı ilə faktiki yaşayış ünvanı bir-birindən fərqli olan işçiləri tapın. 
Bu işçilər arasından yalnız şəxsiyyət vəsiqəsinin etibarlılıq müddəti 2040-cı ildən  tez bitənlərin əlaqə nömrələrini göstərin.
"""
```

```python
payload = {
    "prompt": question
}

headers = {"Content-Type": "application/json"}

try:
    response = requests.post(url, json=payload, headers=headers)
    response.raise_for_status()

    print("Status Code: ", response.status_code)
    print("Response JSON: ")
    print(json.dumps(response.json(), indent=2, ensure_ascii=False))

except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
    if response is not None:
        print(response.text)
```
